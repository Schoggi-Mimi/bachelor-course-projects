```python
N_tp = 56
N_fn = 4
N_fp = 1
N_tn = 180
```

```python
sensitivity = N_tp / (N_tp + N_fn)
specificity = N_tn / (N_tn + N_fp)
sensitivity, specificity
```

```python
tp = [[1, 1]] * N_tp
fn = [[1, 0]] * N_fn
fp = [[0, 1]] * N_fp
tn = [[0, 0]] * N_tn
```

```python
import numpy as np
sample = np.array(tp + fn + fp + tn)
```

```python
from sklearn.metrics import recall_score
def sensitivity(y_true, y_pred):
  return recall_score(y_true=y_true, y_pred=y_pred)
def specificity(y_true, y_pred):
  return recall_score(y_true=1-y_true, y_pred=1-y_pred)
y_true, y_pred = sample.T
sensitivity(y_true, y_pred), specificity(y_true, y_pred)
```

```python
sensitivities = []
specificities = []
for seed in range(10000):
  rng = np.random.default_rng(seed)
  new_sample = rng.choice(sample, size=len(sample), replace=True)
  y_true, y_pred = new_sample.T
  sensitivities.append(sensitivity(y_true, y_pred))
  specificities.append(specificity(y_true, y_pred))
```

```python
print(np.percentile(sensitivities, 2.5), np.percentile(sensitivities, 97.5))
print(np.percentile(specificities, 2.5), np.percentile(specificities, 97.5))
```
