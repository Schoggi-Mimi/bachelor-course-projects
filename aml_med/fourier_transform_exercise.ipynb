```python
import numpy as np
```

```python
phases = np.array([
    [
        (k*n) for k in range(4)
    ]
    for n in range(4)
])/4
```

```python
phases
```

```python
Vn = np.exp(-1j*2*np.pi*phases)
np.around(Vn, 2)
```

```python
phases = np.array([
    [
        (k*n) for k in range(5)
    ]
    for n in range(5)
])/5
```

```python
Vn = np.exp(-1j*2*np.pi*phases)
np.around(Vn, 2)
```

```python
x = np.array([0, 1, 2, 3, 0])
```

```python
np.matmul(Vn, x)
```

```python
from scipy.fft import fft
fft(x)
```

```python
D2 = np.zeros(shape=(5, 5))
for i in range(5):
  D2[i, i] = -2
  D2[i, (i+1) % 5] = +1
  D2[(i+1) % 5, i] = +1
D2
```

```python
diag = np.matmul(np.linalg.inv(Vn), np.matmul(D2, Vn))
```

```python
np.around(diag, 2)
```

```python
Vn_inv = np.exp(+1j*2*np.pi*phases)
id = np.matmul(Vn, Vn_inv)
np.around(id, 2)
```
